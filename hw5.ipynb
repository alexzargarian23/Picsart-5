. Տրված է.
cities = ["Yerevan", "Paris", "London", "Tokyo"]
Փոխարինեք "Paris" տարրը "Rome"-ով և տպեք ստացված list-ը։


```
cities = ["Yerevan", "Paris", "London", "Tokyo"]
for i in range(len(cities)):
    if cities[i] == "Paris":
        cities[i] = "Rome"
print(cities)
```

6. Տրված է.

Տպեք.
առաջին երեք տարրերը,
վերջին երկու տարրերը,

4. Ստեղծեք հետևյալ list-ը.
students = [
   ("Anna", 91),
   ("David", 83),
   ("Mariam", 97)
]
Տպեք յուրաքանչյուր ուսանողի անունն ու գնահատականը հետևյալ ձևաչափով.
Anna -> 91
David -> 83
Mariam -> 97


```
students = [
    ("Anna",91),
    ("David", 83),
    ("Mariam", 97)
]
first , second = students[0]
third, fourth = students[1]
fifth , sixth = students[2]
print(f"{first}->{second}")
print(f"{third}->{fourth}")
print(f"{fifth}->{sixth}")
```
